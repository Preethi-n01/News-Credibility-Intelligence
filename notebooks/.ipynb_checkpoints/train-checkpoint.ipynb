```python
import pandas as pd
import pickle

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
fake = pd.read_csv("../data/Fake.csv")
real = pd.read_csv("../data/True.csv")

fake["label"] = 1
real["label"] = 0

data = pd.concat([fake, real])
data = data[["text", "label"]]

data = data.sample(frac=1).reset_index(drop=True)
```

```python
fake_vectorizer = TfidfVectorizer(stop_words='english', max_features=5000)

X = fake_vectorizer.fit_transform(data["text"])
y = data["label"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

fake_model = LogisticRegression(max_iter=200)
fake_model.fit(X_train, y_train)

y_pred = fake_model.predict(X_test)
print("Fake News Accuracy:", accuracy_score(y_test, y_pred))
```

```python
clickbait = pd.read_csv("../data/clickbait.csv")

click_vectorizer = TfidfVectorizer(stop_words='english', max_features=3000)

X_cb = click_vectorizer.fit_transform(clickbait["headline"])
y_cb = clickbait["label"]

click_model = LogisticRegression(max_iter=200)
click_model.fit(X_cb, y_cb)
```

```python
pickle.dump(fake_model, open("../models/fake_model.pkl", "wb"))
pickle.dump(click_model, open("../models/click_model.pkl", "wb"))

pickle.dump(fake_vectorizer, open("../models/fake_vectorizer.pkl", "wb"))
pickle.dump(click_vectorizer, open("../models/click_vectorizer.pkl", "wb"))

print("Models saved successfully!")
```
